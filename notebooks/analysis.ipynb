# Analysis Renderer

Thin notebook wrapper around the repository graph renderer. Set `RUN_ROOT` to an existing run and execute the render cell.


```python
from pathlib import Path
import importlib
import sys

cwd = Path.cwd().resolve()
ROOT = cwd
for p in [cwd, *cwd.parents]:
    if (p / 'src' / 'graphs' / 'runtime.py').exists():
        ROOT = p
        break
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import src.graphs.runtime as _graphs_runtime
_graphs_runtime = importlib.reload(_graphs_runtime)
render_run = _graphs_runtime.render_run
```

```python
RUN_ROOT = ROOT / 'analysis_out' / 'run-id'
RENDERER_PRESET = 'paper'

if not RUN_ROOT.exists():
    print(f'Set RUN_ROOT to an existing run root. Current value: {RUN_ROOT}')
else:
    result = render_run(RUN_ROOT, preset=RENDERER_PRESET)
    print(result)
```
